```python
import pandas as pd
import numpy as np
data={
    'Name':['anu','ravi','priya','kiran','meena','arjum'],
    'Age':[21,22,None,23,21,None],
    'Salary':[25000,None,30000,28000,None,32000],
    'Branch':['cse','ece','cse',None,'ise','ece'],
    'city':['mysore','banglore','mysore','hubli','banglore','mysore']
}
df=pd.DataFrame(data)
print(df)
```

```python
df["Salary"]=df["Salary"].fillna(df["Salary"].median())
print(df)
```

```python
print(df.isnull())
```

```python
print(df.isnull().sum())
```

```python
print("mean Amputation\n")
df["Age"]=df["Age"].fillna(df['Age'].mean())
print(df)
```

```python
print("mean age",df["Age"].mean())
```

```python
import pandas as pd
import numpy as np
data={
    'Name':['anu','ravi','priya','kiran','meena','arjum'],
    'Age':[21,22,None,23,21,None],
    'Salary':[25000,None,30000,28000,None,32000],
    'Branch':['cse','ece','cse',None,'ise','ece'],
    'city':['mysore','banglore','mysore','hubli','banglore','mysore']
}

df=pd.DataFrame(data)
df['Salary']=df['Salary'].fillna(df["Salary"].median())
print("Median imputation\n")
print(df)
```

```python
data={
    'Name':['anu','ravi','priya','kiran','meena','arjum'],
    'Branch':['cse','ece','cse',None,'ise','ece']
}
df=pd.DataFrame(data)
print("Before forward fill:")
print(df)
df["Branch"]=df['Branch'].ffill()
print(df)
```

```python
from sklearn.preprocessing import LabelEncoder

data={
    'Name':['anu','ravi','priya','kiran','meena','arjum'],
    'Branch':['cse','ece','cse','civil','ise','ece']
}
df=pd.DataFrame(data)
encoder=LabelEncoder()
df['Branch_Encoded']=encoder.fit_transform(df["Branch"])
print(df)
print(encoder.classes_)
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

data={
    'Name':['anu','ravi','priya','kiran','meena','arjum'],
    'Branch':['cse','ece','cse','civil','ise','ece']
}
df=pd.DataFrame(data)
encoded_df=pd.get_dummies(df,columns=["Branch"])
print(encoded_df)
```
